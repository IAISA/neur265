```python
#importing necessary packages
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
#imports csv files as numopy arrays
v_rest_sst = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv')
v_rest_pvalb = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv')
```

```python
#finds mean, t-test, plots histogram, and labels axes
#mean
mean_sst = np.mean(v_rest_sst)
mean_pvalb = np.mean(v_rest_pvalb)

#ttest
tstat, pvalue = stats.ttest_ind(v_rest_sst, v_rest_pvalb)

#histograms
if pvalue < 0.05:
  fig, ax = plt.subplots(1,2, figsize = (10,4))

  ax[0].hist(v_rest_sst, bins=20)
  ax[0].set_title('sst hisogram')
  ax[0].set_xlabel('membrane potential')
  ax[0].set_ylabel('frequency')

  ax[1].hist(v_rest_pvalb, bins=20)
  ax[1].set_title('pvalb hisogram')
  ax[1].set_xlabel('membrane potential')
  ax[1].set_ylabel('frequency')

  plt.tight_layout()
  plt.show()
```

```python
#defines a function called vrestHist()
def vrestHist():
  mean_sst = np.mean(v_rest_sst)
  mean_pvalb = np.mean(v_rest_pvalb)

  #ttest
  tstat, pvalue = stats.ttest_ind(v_rest_sst, v_rest_pvalb)

  #histograms
  if pvalue < 0.05:
    fig, ax = plt.subplots(1,2, figsize = (10,4))

    ax[0].hist(v_rest_sst, bins=20)
    ax[0].set_title('sst hisogram')
    ax[0].set_xlabel('membrane potential')
    ax[0].set_ylabel('frequency')

    ax[1].hist(v_rest_pvalb, bins=20)
    ax[1].set_title('pvalb hisogram')
    ax[1].set_xlabel('membrane potential')
    ax[1].set_ylabel('frequency')

    plt.tight_layout()
    plt.show()
```

```python
#call function
vrestHist()
```

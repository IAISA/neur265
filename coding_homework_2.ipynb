```python
#step 2, all the imports
import numpy as np
import matplotlib.pyplot as plt
import statistics as st
```

```python
#step 3, imports data asa arrays
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
v_rest1 = np.loadtxt(url1)

url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
v_rest2 = np.loadtxt(url2)
```

```python
#step 4, concatenates sst and pvald into matrix
#i got 500 columns so I limited to 13
v_rest = np.stack((v_rest1[:13], v_rest2[:13]))
shape = (2,13)
print(v_rest.shape)
```

```python
#10, mean of the values in teh first row of v_rest and assigning the mean
v_rest_mean_sst = v_rest[0].mean()
v_rest_mean_pvalb = v_rest[1].mean()

print(v_rest_mean_sst)
print(v_rest_mean_pvalb)
```

###Question 11: the mean resting membrane potential of the first row (sst) is -69.102 mV, while the mean of pvalb is -67.835 mV. The first row is slightly lower, more negative than teh other. The difference in this case is about 1.267 mV.

```python
#12, import scipy package
import scipy.stats as stats
```

```python
#13, comapring two means, using t-test
statistic, pvalue = stats.ttest_ind(v_rest[0], v_rest[1])
print('t_stat', statistic)
print('p value:', pvalue)
```

### Question 14: the t-test produced a t-stat of -0.466 and a p-value of 0.646. Since the p_value is greater tha 0.05 the difference between the two means is not statistically significant. Which means that the means are not significantly different.

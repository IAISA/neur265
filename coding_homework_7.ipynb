```python
#import the libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#import files
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv'

acc_hits = np.loadtxt(url1, delimiter = ',')
acc_fa = np.loadtxt(url2, delimiter = ',')
```

```python
#prints the shape of each array
print(acc_hits.shape)
print(acc_fa.shape)
```

```python
#time variable that has the same length as the two oyher vars
time = np.linspace(0, 4, acc_hits.shape[1])
```

```python
#the sampling rate of teh data trial
sampling = 8001/4
print(sampling)
```

```python
#plots teh first trials of acc_hots and acc_fa
plt.figure(figsize=(10,6))

plt.subplot(2,1,1)
plt.plot(time, acc_hits[0])
plt.xlabel('time')
plt.ylabel('acc_hits')

plt.subplot(2,1,2)
plt.plot(time, acc_fa[0])
plt.xlabel('time')
plt.ylabel('acc_fa')

plt.show()
```

Question 9: for acc_hits the post touch oscillations seem to happen every .2-.25 seconds, most likely meaning a dominant frequency around 4-5Hz. for acc_fa it was broader, more slower oscillation meaning a more dodminant osciliation as well at 1-2 Hz. They do differ overall because acc_hits showed faster oscillations than acc_fa.

```python
#compare LFPs
erp_hits = np.mean(acc_hits, axis=0)

plt.figure(figsize=(10,4))
plt.plot(time, erp_hits, color='blue', linewidth=1.5)

plt.axvline(x=2, color='black', linewidth=1.5, label='screen touch')

plt.axhline(y=0, color='black', linewidth=1, linestyle='--')

plt.xlabel('time')
plt.ylabel('LFP amplitude')
plt.legend()
plt.show()
```

Question 11: I think I do see evidence of consistent activity in the LFP surrounding the black bar because before the 2 second mark the signal is flat and neat zero. and right after the black bar the ERP shows a sharp negative deflection then a more positive peak.

```python
#repeat of step 10 for acc_fa

erp_fa = np.mean(acc_fa, axis = 0)

plt.figure(figsize=(10,4))
plt.plot(time, erp_hits, color='blue', linewidth=1.5, label='acc_hits')
plt.plot(time, erp_fa, color='red', linewidth=1.5, label='acc_fa')

plt.axvline(x=2, color='black', linewidth=1.5, label='screen touch')

plt.axhline(y=0, color='black', linewidth=1, linestyle='--')

plt.xlabel('time')
plt.ylabel('LFP amplitude')
plt.legend()
plt.show()
```

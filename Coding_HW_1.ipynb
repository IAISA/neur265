```python
#4.  importing .csv files as pandas data frames
import pandas as pd
url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
pandas_data1 = pd.read_csv(url)

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
pandas_data2 = pd.read_csv(url)
```

```python
#5. converts data to a list
data1 = pandas_data1.values.tolist()
data2 = pandas_data2.values.tolist()
```

```python
#6. flattening data
import numpy as np

data1 = np.squeeze(data1)
data2 = np.squeeze(data2)
```

```python
#7. prints the lngth of data 1
print(len(data1))

#8. prints the 3,4,5,6th element
print(data2[2:6])

#9. prints second to last and last elements of data 1
print(data1[-2:])

#10. prints the type
print(type(data2))
```

```python
#11. make histogram
import matplotlib.pyplot as plt

plt.hist(data1)
```

```python
#12. add labels
plt.hist(data1)
plt.xlabel('reseting membrane potential')
plt.ylabel('# of cells')
plt.title('histogram of data1 - SST values')
```

```python
#same thing for data 2

plt.hist(data2)
plt.xlabel('reseting membrane potential')
plt.ylabel('# of cells')
plt.title('histogram of data2 - PV cells')
```

```python
#import
import numpy as np
import matplotlib.pyplot as plt
```

```python
#import ffiles as numpy arrays
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv'
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv'
url3 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
url4 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv'

spikes_left_trial1 = np.loadtxt(url1, delimiter = ',')
spikes_right_trial1 = np.loadtxt(url2, delimiter = ',')
spikes_left_trial2 = np.loadtxt(url3, delimiter = ',')
spikes_right_trial2 = np.loadtxt(url4, delimiter = ',')
```

```python
#length of each array
print('Left Triall 1:', len(spikes_left_trial1))
print('Right Triall 1:', len(spikes_right_trial1))
print('Left Triall 2:', len(spikes_left_trial2))
print('Right Triall 2:', len(spikes_right_trial2))
```

```python
#computing firing rates
seconds = 30
rateL1 = len(spikes_left_trial1)/seconds
rateR1 = len(spikes_right_trial1)/seconds
rateL2 = len(spikes_left_trial2)/seconds
rateR2 = len(spikes_right_trial2)/seconds

print('left trial 1:', rateL1)
print('right trial 1:', rateR1)

print('left trial 2:', rateL2)
print('right trial 2:', rateR2)
```

Question 7: It seems like the neuron does not consistently fire more for one direction. It fires the most during the right turn in tral 2 and it is not too far from the left trial 1 rate, so this means there is no clear prefernce for left or right turns.

```python
#making the two sublots
plt.figure(figsize=(10,5))

#trial 1
plt.subplot(2,1,1)
plt.scatter(spikes_left_trial1, [1]*len(spikes_left_trial1), label = 'left', color = 'red')
plt.scatter(spikes_right_trial1, [2]*len(spikes_right_trial1), label = 'right', color = 'blue')
plt.yticks([1,2], ['left', 'right'])
plt.xlabel('time')
plt.ylabel('trial 1')

#trial2
plt.subplot(2,1,2)
plt.scatter(spikes_left_trial2, [1]*len(spikes_left_trial2), label = 'left', color = 'red')
plt.scatter(spikes_right_trial2, [2]*len(spikes_right_trial2), label = 'right', color = 'blue')
plt.yticks([1,2], ['left', 'right'])
plt.xlabel('time')
plt.ylabel('trial 1')

plt.show()
```

Question 9: thee neauron seems to be more bursty in the right turn trials whereas the firing during the left turn trials looks more spread out and regular

```python
#calculate ISI
isiLeft1 = np.diff(spikes_left_trial1)
isiRight1 = np.diff(spikes_right_trial1)
isiLeft2 = np.diff(spikes_left_trial2)
isiRight2 = np.diff(spikes_right_trial2)
```

```python
#make 4 histograms
bins = np.arange(0, 2.01, 0.01)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.hist(isiLeft1,bins=bins)
plt.title('ISI left trial 1')

plt.subplot(2,2,2)
plt.hist(isiRight1,bins=bins)
plt.title('ISI right trial 1')

plt.subplot(2,2,3)
plt.hist(isiLeft2,bins=bins)
plt.title('ISI left trial 2')

plt.subplot(2,2,4)
plt.hist(isiRight2,bins=bins)
plt.title('ISI right trial 2')

plt.show()
```

Question 12: the neuron does show bursty behavior, as many ISIs are clustered near 0 seconds. The burstiness seems to be strong during the right turn trials, espeially in trial2, where there are a higher concentration of short ISIs compared to the left turn trials.

```python
#calculate the firing rate across time for each array
bins = np.arange(0,31,1)

#histo counts
rateL1, _ = np.histogram(spikes_left_trial1, bins=bins)
rateR1, _ = np.histogram(spikes_right_trial1, bins=bins)
rateL2, _ = np.histogram(spikes_left_trial2, bins=bins)
rateR2, _ = np.histogram(spikes_right_trial2, bins=bins)

time = bins[:-1]
plt.figure(figsize=(10,5))

#trial1
plt.subplot(2,1,1)
plt.plot(time, rateL1, color='red', label='left')
plt.plot(time, rateR1, color='blue', label='right')
plt.title('trial 1')
plt.xlabel('time')
plt.ylabel('firing rate')

#trial2
plt.subplot(2,1,2)
plt.plot(time, rateL2, color='red', label='left')
plt.plot(time, rateR2, color='blue', label='right')
plt.title('trial 2')
plt.xlabel('time')
plt.ylabel('firing rate')

plt.show()
```
